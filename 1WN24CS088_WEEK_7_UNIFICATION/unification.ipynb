```python
print("DEEPTHI GIRISH - 1WN24CS088")
def unify(x, y, subs):
    if x == y:
        return subs

    if isinstance(x, str) and x.islower():
        if x in subs:
            return unify(subs[x], y, subs)
        if occurs(x, y, subs):
            return None
        subs[x] = y
        return subs

    if isinstance(y, str) and y.islower():
        return unify(y, x, subs)

    if isinstance(x, tuple) and isinstance(y, tuple):
        if x[0] != y[0] or len(x[1]) != len(y[1]):
            return None

        for a, b in zip(x[1], y[1]):
            subs = unify(a, b, subs)
            if subs is None:
                return None
        return subs

    return None


def occurs(var, term, subs):
    if var == term:
        return True
    if isinstance(term, tuple):
        return any(occurs(var, t, subs) for t in term[1])
    if term in subs:
        return occurs(var, subs[term], subs)
    return False


def parse(expr):
    expr = expr.strip()
    if "(" not in expr:
        return expr

    name, args = expr.split("(", 1)
    args = args[:-1]
    parts = []
    depth = 0
    current = ""

    for ch in args:
        if ch == "," and depth == 0:
            parts.append(parse(current))
            current = ""
        else:
            current += ch
            if ch == "(":
                depth += 1
            elif ch == ")":
                depth -= 1

    if current.strip():
        parts.append(parse(current))

    return (name.strip(), parts)


def display(term):
    if isinstance(term, tuple):
        return term[0] + "(" + ", ".join(display(t) for t in term[1]) + ")"
    return term
e1 = input("Enter first expression: ")
e2 = input("Enter second expression: ")

result = unify(parse(e1), parse(e2), {})

if result is None:
    print("Unification Failed")
else:
    print("Unification Successful")
    print("Substitutions:")
    for var, value in result.items():
        print(var, "=", display(value))

    print("Unified Expression:", display(parse(e1)))
```
