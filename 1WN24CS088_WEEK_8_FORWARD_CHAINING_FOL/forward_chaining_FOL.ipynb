```python

import re
print("DEEPTHI GIRISH-1WN24CS088")
def parse_atom(text):
    match = re.match(r'\s*(\w+)\((.*?)\)\s*', text)
    if not match:
        return None
    return (
        match.group(1),
        tuple(x.strip() for x in match.group(2).split(","))
    )
def is_variable(value):
    return value[0].islower()
def unify(pattern, fact, substitution):
    if pattern[0] != fact[0] or len(pattern[1]) != len(fact[1]):
        return None
    sub = substitution.copy()
    for p, f in zip(pattern[1], fact[1]):
        if is_variable(p):
            if p in sub and sub[p] != f:
                return None
            sub[p] = f
        elif p != f:
            return None
    return sub
def display(atom):
    return atom[0] + "(" + ", ".join(atom[1]) + ")"
def forward_chaining(facts, rules, query):
    known = set(facts)
    step = 1
    print("\nINITIAL FACTS:")
    for fact in facts:
        print(display(fact))
    changed = True
    print("\nFORWARD CHAINING PROCESS")
    while changed:
        changed = False
        for conditions, conclusion, rule_no, rule_text in rules:
            substitutions = [{}]
            for condition in conditions:
                next_substitutions = []
                for sub in substitutions:
                    for fact in sorted(known):
                        result = unify(condition, fact, sub)
                        if result is not None:
                            next_substitutions.append(result)
                substitutions = next_substitutions
            for sub in substitutions:
                matched_facts = [
                    (condition[0],
                     tuple(sub.get(x, x) for x in condition[1]))
                    for condition in conditions
                ]
                args = tuple(
                    sub.get(x, x) for x in conclusion[1]
                )
                inferred = (conclusion[0], args)
                if inferred not in known:
                    print("\nSTEP", step)
                    print("Rule Applied:", rule_no)
                    print("Rule:", rule_text)

                    print("Matching Facts:")
                    for fact in matched_facts:
                        print("  ", display(fact))
                    print("New Fact Inferred:", display(inferred))
                    known.add(inferred)
                    changed = True
                    step += 1
                    if inferred == query:
                        print("\nGOAL REACHED!")
                        print("Goal:", display(query))
                        return True
    print("\nGOAL NOT PROVED")
    return False
print("FIRST ORDER LOGIC - FORWARD CHAINING")
facts = []
print("\nEnter FACTS (Type END to finish)")
while True:
    line = input("> ").strip()
    if line.upper() == "END":
        break
    atom = parse_atom(line)
    if atom:
        facts.append(atom)
rules = []
print("\nEnter RULES")
print("Format: IF A(x) AND B(x) THEN C(x)")
print("Type END to finish")
rule_no = 1
while True:
    line = input("> ").strip()
    if line.upper() == "END":
        break
    if "THEN" not in line.upper():
        print("Invalid rule")
        continue
    parts = re.split(r'\bTHEN\b', line, flags=re.IGNORECASE)
    left = re.split(
        r'\bIF\b', parts[0], flags=re.IGNORECASE
    )[-1]
    conditions = [
        parse_atom(x.strip())
        for x in re.split(r'\bAND\b', left, flags=re.IGNORECASE)
    ]
    conclusion = parse_atom(parts[1].strip())
    if all(conditions) and conclusion:
        rules.append((conditions, conclusion, rule_no, line))
        rule_no += 1
    else:
        print("Invalid rule")
print("\nEnter GOAL to prove:")
query = parse_atom(input("> "))
if query:
    result = forward_chaining(facts, rules, query)
    if result:
        print("\nFINAL RESULT: SUCCESS")
    else:
        print("\nFINAL RESULT: FAILURE")
else:
    print("Invalid goal")
```
